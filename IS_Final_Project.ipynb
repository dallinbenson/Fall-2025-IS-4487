```python
import pandas as pd
import re
from io import StringIO

file_name_demographics = 'healthinsurancedemographictables2023.csv'

# Read the file line by line to get specific header and data lines
# It's important to re-read to ensure we have the raw lines as a list for precise slicing.
with open(file_name_demographics, 'r', encoding='utf-8') as f:
    lines = f.readlines()

# --- 1. Process Percentage Data Block ---
# Data lines for percentages (lines 9 to 13, inclusive)
percentage_data_raw = lines[9:14]

# Manually defined column names based on detailed inspection of lines 5, 6, and 7
column_names_percent = [
    "Age_group",
    "Private_coverage_percent",
    "Public_coverage_percent",
    "Uninsured_total_percent",
    "Uninsured_less_than_1_year_percent",
    "Uninsured_1_year_or_more_percent",
    "Uninsured_unknown_duration_percent"
]

# Clean up data lines and prepare for DataFrame creation
# Remove quotes, replace en-dash, and split by multiple spaces
cleaned_percentage_data_list = []
for line in percentage_data_raw:
    cleaned_line = line.strip().strip('"').replace('–', '-') # Replace en-dash with hyphen
    # Use re.split to handle varied whitespace between columns, ensuring no split within data points like '61.3 (60.4-62.1)'
    parts = re.split(r'\s{2,}', cleaned_line)
    cleaned_percentage_data_list.append(parts)

# Create DataFrame for percentages
df_demographics_percent = pd.DataFrame(cleaned_percentage_data_list, columns=column_names_percent)

# --- 2. Process Numbers in Millions Data Block ---
# Data lines for numbers (lines 17 to 21, inclusive)
numbers_data_raw = lines[17:22]

# Manually defined column names based on detailed inspection of lines 5, 6, and 15
column_names_numbers = [
    "Age_group",
    "Private_coverage_millions",
    "Public_coverage_millions",
    "Uninsured_total_millions",
    "Uninsured_less_than_1_year_millions",
    "Uninsured_1_year_or_more_millions",
    "Uninsured_unknown_duration_millions"
]

# Clean up data lines and prepare for DataFrame creation
cleaned_numbers_data_list = []
for line in numbers_data_raw:
    cleaned_line = line.strip().strip('"')
    parts = re.split(r'\s{2,}', cleaned_line)
    cleaned_numbers_data_list.append(parts)

# Create DataFrame for numbers in millions
df_demographics_numbers = pd.DataFrame(cleaned_numbers_data_list, columns=column_names_numbers)

# --- 3. Initial Data Cleaning and Type Conversion ---

# Function to extract the primary numerical value from a string like "61.3 (60.4-62.1)"
def extract_numeric_value(text):
    if pd.isna(text) or not isinstance(text, str):
        return None
    match = re.match(r'(\d+\.?\d*)', text)
    return float(match.group(1)) if match else None

# Apply cleaning and type conversion to percentage DataFrame
for col in df_demographics_percent.columns:
    if col != "Age_group":
        df_demographics_percent[col] = df_demographics_percent[col].apply(extract_numeric_value)

# Convert numerical columns in numbers DataFrame to float
for col in df_demographics_numbers.columns:
    if col != "Age_group":
        df_demographics_numbers[col] = pd.to_numeric(df_demographics_numbers[col], errors='coerce')

# --- 4. Display Results for Initial Inspection ---
print("\n--- DataFrame for Percentage Data (df_demographics_percent) ---")
print("First 5 rows:")
print(df_demographics_percent.head())
print("\nShape:", df_demographics_percent.shape)
print("\nInfo:")
df_demographics_percent.info()

print("\n\n--- DataFrame for Numbers in Millions Data (df_demographics_numbers) ---")
print("First 5 rows:")
print(df_demographics_numbers.head())
print("\nShape:", df_demographics_numbers.shape)
print("\nInfo:")
df_demographics_numbers.info()
```

### Handling Outliers (Identification and Strategy)


```python
# Using IQR method for outlier detection

# --- For temp_df_states ---
print("\n--- Outlier Detection in temp_df_states (using IQR) ---")
numerical_cols_states = temp_df_states.select_dtypes(include=np.number).columns.tolist()

for col in numerical_cols_states:
    Q1 = temp_df_states[col].quantile(0.25)
    Q3 = temp_df_states[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = temp_df_states[(temp_df_states[col] < lower_bound) | (temp_df_states[col] > upper_bound)]
    if not outliers.empty:
        print(f"Column '{col}' has {len(outliers)} outliers. Example outliers:\n{outliers[['state', col]].head()}")
    # For this analysis, we'll primarily identify and note outliers.
    # Capping or removal would depend on the specific modeling strategy and data distribution.

# --- For df_demographics ---
print("\n--- Outlier Detection in df_demographics (using IQR) ---")
numerical_cols_demographics = df_demographics.select_dtypes(include=np.number).columns.tolist()

for col in numerical_cols_demographics:
    Q1 = df_demographics[col].quantile(0.25)
    Q3 = df_demographics[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df_demographics[(df_demographics[col] < lower_bound) | (df_demographics[col] > upper_bound)]
    if not outliers.empty:
        print(f"Column '{col}' has {len(outliers)} outliers. Example outliers:\n{outliers[['Age_group', col]].head()}")
    # Similar to temp_df_states, we'll identify outliers. Further action might be needed based on model choice.
```

## 6.4 Feature Engineering

```python
import numpy as np

# --- Cleaning and converting columns in temp_df_states ---

# Function to convert percentage strings to float
def convert_percent_to_float(df, column_name):
    # Ensure the column exists and is of object type before proceeding
    if column_name in df.columns and df[column_name].dtype == 'object':
        df[column_name] = df[column_name].str.replace('%', '', regex=False).astype(float)
    return df

# Function to convert currency strings to float
def convert_currency_to_float(df, column_name):
    if column_name in df.columns and df[column_name].dtype == 'object':
        df[column_name] = df[column_name].str.replace('$', '', regex=False).astype(float)
    return df

# Apply conversions to temp_df_states
# These conversions are necessary before some calculations if the columns are still strings
temp_df_states = convert_percent_to_float(temp_df_states, 'uninsured_rate_2010')
temp_df_states = convert_percent_to_float(temp_df_states, 'uninsured_rate_2015')
# 'uninsured_rate_change_20102015' might also be a string percentage, ensure conversion
temp_df_states = convert_percent_to_float(temp_df_states, 'uninsured_rate_change_20102015')
temp_df_states = convert_currency_to_float(temp_df_states, 'average_monthly_tax_credit_2016')

print("--- Data types of temp_df_states after cleaning relevant columns ---")
print(temp_df_states[['uninsured_rate_2010', 'uninsured_rate_2015', 'uninsured_rate_change_20102015', 'average_monthly_tax_credit_2016']].dtypes)


# --- Feature Engineering for temp_df_states ---

# 1. Calculate the absolute difference in uninsured rate
temp_df_states['uninsured_rate_difference_calculated'] = temp_df_states['uninsured_rate_2010'] - temp_df_states['uninsured_rate_2015']

# 2. Marketplace Coverage Ratio (Marketplace Coverage / Employer Coverage)
temp_df_states['marketplace_per_employer_coverage_ratio'] = np.where(
    temp_df_states['employer_health_insurance_coverage_2015'] != 0,
    temp_df_states['marketplace_health_insurance_coverage_2016'] / temp_df_states['employer_health_insurance_coverage_2015'],
    np.nan  # Handle division by zero
)

# 3. Medicaid Enrollment Growth Rate
temp_df_states['medicaid_enrollment_growth_rate'] = np.where(
    (temp_df_states['medicaid_enrollment_2013'].notna()) & (temp_df_states['medicaid_enrollment_2013'] != 0),
    temp_df_states['medicaid_enrollment_change_20132016'] / temp_df_states['medicaid_enrollment_2013'],
    np.nan # Handle division by zero or NaN in denominator
)

print("\n--- First 5 rows of temp_df_states with new features ---")
print(temp_df_states[['state', 'uninsured_rate_2010', 'uninsured_rate_2015', 'uninsured_rate_difference_calculated', 'marketplace_per_employer_coverage_ratio', 'medicaid_enrollment_growth_rate']].head())
```

```python
# --- Feature Engineering for df_demographics ---

# 1. Private to Public Coverage Ratio (Percentage)
df_demographics['private_public_coverage_ratio_percent'] = np.where(
    df_demographics['Public_coverage_percent'] != 0,
    df_demographics['Private_coverage_percent'] / df_demographics['Public_coverage_percent'],
    np.nan # Handle division by zero
)

# 2. Ratio of long-term uninsured to short-term uninsured (Percentage)
df_demographics['uninsured_duration_ratio_percent'] = np.where(
    df_demographics['Uninsured_less_than_1_year_percent'] != 0,
    df_demographics['Uninsured_1_year_or_more_percent'] / df_demographics['Uninsured_less_than_1_year_percent'],
    np.nan # Handle division by zero
)

print("\n--- First 5 rows of df_demographics with new features ---")
print(df_demographics[['Age_group', 'Private_coverage_percent', 'Public_coverage_percent', 'private_public_coverage_ratio_percent', 'Uninsured_1_year_or_more_percent', 'Uninsured_less_than_1_year_percent', 'uninsured_duration_ratio_percent']].head())
```

### Summary of 6.4 Feature Engineering

In this subsection, we performed feature engineering to derive new, potentially more insightful, variables from our existing datasets:

**For the `states` dataset (`temp_df_states`):**
1.  **Data Type Conversion**: We first ensured that several string-based columns containing percentages (e.g., `uninsured_rate_2010`, `uninsured_rate_2015`, `uninsured_rate_change_20102015`) and currency (`average_monthly_tax_credit_2016`) were converted into numerical (float) types. This is crucial for mathematical operations and model compatibility.
2.  **`uninsured_rate_difference_calculated`**: A new feature was calculated to represent the absolute difference in uninsured rates between 2010 and 2015 for each state, providing a direct measure of change.
3.  **`marketplace_per_employer_coverage_ratio`**: This ratio was created to understand the reliance on marketplace health insurance relative to employer-sponsored health insurance within each state.
4.  **`medicaid_enrollment_growth_rate`**: Calculated as the change in Medicaid enrollment from 2013 to 2016 relative to the 2013 enrollment, indicating the growth or decline in Medicaid beneficiaries for each state.

**For the `demographics` dataset (`df_demographics`):**
1.  **`private_public_coverage_ratio_percent`**: This feature represents the ratio of private health insurance coverage percentage to public health insurance coverage percentage by age group, offering insights into the dominance of one coverage type over the other.
2.  **`uninsured_duration_ratio_percent`**: This ratio compares the percentage of people uninsured for one year or more to those uninsured for less than one year, helping to identify trends in long-term vs. short-term uninsurance by age group.

These new features aim to capture deeper relationships and patterns in the data, enhancing the predictive power for subsequent modeling stages.

# 7. Model Development

## 7.1 Model Selection & Rationale

For predicting health insurance premiums and coinsurance rates, which is a regression problem, we will consider testing the following models:

1.  **Linear Regression**:
    *   **Description**: A fundamental algorithm that models the relationship between a dependent variable (the premium/coinsurance) and one or more independent variables (features) by fitting a linear equation to the observed data. It assumes a linear relationship between features and the target variable.
    *   **Rationale**: Linear regression serves as an excellent baseline model due to its simplicity and interpretability. It allows us to understand the direct, additive influence of each feature on the cost. If the relationships are largely linear, it can provide effective predictions and clear insights into key cost drivers.

2.  **Random Forest Regressor**:
    *   **Description**: An ensemble learning method that builds multiple decision trees during training and outputs the mean prediction of the individual trees. It is robust to overfitting and can handle both linear and non-linear relationships as well as a large number of features.
    *   **Rationale**: Health insurance costs are likely influenced by complex, non-linear interactions between various demographic, coverage, and market factors. Random Forest can effectively capture these intricate relationships, handle different data types (numerical and categorical), and generally provides higher accuracy than simpler models without significant tuning.



## 7.2 Train/Test Split or Cross-Validation Approach

For evaluating the performance and generalization capabilities of our models, we need a robust validation strategy. We will consider two common approaches:

1.  **Train/Test Split**:
    *   **Description**: The dataset is divided into two mutually exclusive subsets: a training set and a testing set (e.g., 80% for training, 20% for testing). The model is trained exclusively on the training data and then evaluated on the unseen testing data. This provides an estimate of how the model will perform on new, unobserved data.
    *   **Rationale**: This approach is straightforward to implement and interpret. It's often suitable for large datasets where having a substantial test set doesn't significantly reduce the training data's representativeness. It offers a quick way to get an initial performance estimate.

2.  **K-Fold Cross-Validation**:
    *   **Description**: The dataset is divided into *k* equal-sized folds. The model is trained *k* times; in each iteration, one fold is used as the testing set, and the remaining *k-1* folds are used as the training set. The performance metrics are averaged across all *k* iterations.
    *   **Rationale**: Cross-validation provides a more reliable estimate of model performance and reduces the variance of the performance estimate compared to a single train/test split, especially for smaller to moderately sized datasets. It ensures that every data point is used for both training and testing at least once, making better use of the available data and providing a more robust measure of generalization capability.

### Chosen Approach

Given the nature of the data and the goal of achieving reliable predictions, we will primarily utilize **K-Fold Cross-Validation** for evaluating our models. This approach will help in obtaining a more stable and less biased estimate of model performance, ensuring our models generalize well to new health insurance data. A typical value for *k* (e.g., 5 or 10) will be chosen based on dataset size and computational resources.
