## Getting data into Pandas

In this case we will load data from the statsmodels.org library

MTCARS is a traditional dataset used for learning.



```python
mtcars = sm.datasets.get_rdataset("mtcars", "datasets", cache=True).data
df = pd.DataFrame(mtcars)
df['carmodel'] = df.index
print(df)
```

## Preview Data

Use the following data dictionary to understand the attributes available


# Create basic charts

- Line chart
- Bar chart
- Histogram
- Box plot
- Scatter plot




```python
#Line chart
df.plot(x='carmodel', y='mpg', kind='line', title="MPG by Car")
```

```python
#Bar chart
df.plot(x='carmodel', y='mpg', kind='bar', title="MPG by Car")
```

```python
#Histogram
df['mpg'].plot(kind='hist', bins=10, title="MPG Distribution")
```

```python
#Boxplot
df['mpg'].plot(kind='box', title = 'Miles Per Gallon')
```

```python
#Scatter plot
df.plot.scatter(x='mpg',y='hp',c='wt')
```

```python
#Pie chart
df['am'].value_counts().plot(kind='pie', autopct='%1.1f%%', title="Distribution of Transmission Types")
```

```python
import matplotlib.pyplot as plt
# Extract manufacturer from carmodel (assuming the first word is the manufacturer)
df['manufacturer'] = df['carmodel'].apply(lambda x: x.split(' ')[0])

# Get unique manufacturers and assign colors using a colormap
manufacturers = df['manufacturer'].unique()
colors = plt.cm.get_cmap('tab20', len(manufacturers))
color_map = {manufacturer: colors(i) for i, manufacturer in enumerate(manufacturers)}

# Create a list of colors for each car model based on its manufacturer
bar_colors = df['manufacturer'].map(color_map)

# Create a bar chart of cars and HP, colored by manufacturer
df.plot(x='carmodel', y='hp', kind='bar', color=bar_colors, title="HP by Car Model (Colored by Manufacturer)", legend=False)
```

```python
#Scatter plot
df.plot.scatter(x='wt', y='mpg', title='Weight vs. Miles Per Gallon')
```

```python
#Bar chart
df.groupby('cyl')['mpg'].mean().plot(kind='bar', title='Average MPG by Number of Cylinders')
```

```python
#Histogram
df['wt'].plot(kind='hist', title='Distribution of Weight')
```
